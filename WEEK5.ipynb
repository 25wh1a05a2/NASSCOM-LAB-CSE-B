```python
import seaborn as sns
tips=sns.load_dataset("tips")
tips.head()
```

```python
import matplotlib.pyplot as plt
plt.hist(
    tips["total_bill"],
    bins=20,
    edgecolor="black"
)
```

```python
import matplotlib.pyplot as plt
counts=tips["smoker"].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90
)
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)
```

```python
import matplotlib.pyplot as plt
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)
```

```python
import matplotlib.pyplot as plt
sns.boxplot(
    data=tips,
    x="sex",
    y="total_bill"
)
```
