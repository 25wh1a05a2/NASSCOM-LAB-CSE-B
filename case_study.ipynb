```python
battery_data=[
    [0,32,400,20,95],
    [1,35,395,22,88],
    [2,38,390,25,80],
    [3,40,385,28,70],
    [4,42,380,30,60],
    [5,45,375,32,48],
]
print("Battery sensor readings:")
for data in battery_data:
  print("Time:",data[0],"h")
  print("Temp:",data[1],"oc")
  print("voltage:",data[2],"v")
  print("current:",data[3],"A")
  print("charge:",data[4],"%")
  for data in battery_data:
    temp=data[1]
    charge=data[4]
    if temp>40:
      print("Time:",data[0],":","High temp Alert")
    else:
      print("Time:",data[0],":","Temp normal")
    if charge<50:
      print("Time:",data[0],":","Low battery alert")
    else:
      print("Time:",data[0],":","Battery charge normal")
  def battery_status(charge):
    if charge>70:
        return "High"
    elif 50<=charge<=70:
          return "Medium"
    else:
      return "Low"
    print("\nBattery status")
    for data in battery_data:
         print("Time",data[0],":",battery_status(data[4]))
```

```python
import numpy as np
A=np.array([[2,1],
           [1,3]])
X=np.array([[40],
           [25]])
Y=np.dot(A,X)
print("Y:")
print(Y)
print("\nTranspose of A:")
print(A.T)
print("\nDeterminant of A:")
print(np.linalg.det(A))
print("\nInverse of A:")
print(np.linalg.inv(A))
B=np.array([[100],
            [120]])
solution=np.linalg.solve(A,B)
print("\nSolution:")
print("x:",solution[0][0])
print("y:",solution[1][0])
```

```python
import sympy as sp
t=sp.symbols('t')
c=100-2*t-sp.Rational(1,2)*t**2
dcdt=sp.diff(c,t)
print("charge function:",c)
print("Rate of change dc/dt:",dcdt)
print(dcdt)
rate_2=dcdt.subs(t,2)
rate_5=dcdt.subs(t,5)
print("Rate of charge at t=2:",rate_2,"% per hour")
print("Rate of charge at t=5:",rate_5,"% per hour")
if abs(rate_5)>abs(rate_2):
  print("The battery is losing charge at an increasing rate")
else:
  print("The battery is losing charge at a decreasing rate")
```
