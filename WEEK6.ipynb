Categorical vs. Numerical Data: Categorical represents labels/groups; Numerical represents measurable quantities.   

Discrete vs. Continuous Data: Discrete consists of distinct, countable values; Continuous represents measurable values on a continuum.

Mean: Sum of values divided by count.

Median: Middle value in sorted order.

When Median is preferable: In skewed distributions or when outliers exist.

Mode: Most frequent value.

Range: Max−Min.

Variance: Average squared deviation from the mean.

Standard Deviation: Square root of variance root(
Variance)

IQR: Range of middle 50% (Q3−Q1).

Skewness: Measure of distribution asymmetry.

Positive vs. Negative Skewness: Positive has a right-side tail (Mean>Median); Negative has a left-side tail (Mean<Median).

Outliers & Mean: Outliers pull the mean toward extreme values.

Why Median is resilient: Median depends on position/order rather than absolute values.

High Std Dev / Small IQR: High Std Dev means high spread from the mean; Small IQR means tight clustering around the middle 50%.

Mean & Median in Positively Skewed Distribution: Mean>Median.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
np.random.seed(42)
df = sns.load_dataset('tips')
print('Loaded tips dataset:', df.shape)
df.head()
```

```python
print(df.dtypes)
numerical   = df.select_dtypes(include='number').columns.tolist()
categorical = df.select_dtypes(exclude='number').columns.tolist()
print('\nNumerical  :', numerical)
print('Categorical:', categorical)
```

```python
col = df['total_bill']
print('Mean   :', round(col.mean(), 2))
print('Median :', round(col.median(), 2))
print('Mode   :', round(col.mode()[0], 2))
```

```python
print('mean > median ?', col.mean() > col.median(), '-> right-skewed')

plt.figure(figsize=(7, 3.5))
sns.histplot(col, kde=True)
plt.axvline(col.mean(),   color='red',   label='mean')
plt.axvline(col.median(), color='green', label='median')
plt.legend(); plt.title('total_bill: mean vs median'); plt.show()
```

```python
tip=df['tip']
print('Mean   :', round(tip.mean(), 2))
print('Median :', round(tip.median(), 2))
print('Mode   :', round(tip.mode()[0], 2))


print('mean > median ?', tip.mean() > tip.median(), '-> right-skewed')
plt.figure(figsize=(7, 3.5))
sns.histplot(tip, kde=True)
plt.axvline(tip.mean(),   color='red',   label='mean')
plt.axvline(tip.median(), color='green', label='median')
plt.legend(); plt.title('tip: mean vs median'); plt.show()
```

```python

col = df['total_bill']
print('Range :', round(col.max() - col.min(), 2))
print('Var   :', round(col.var(), 2))
print('Std   :', round(col.std(), 2))
```

```python
q1, q3 = col.quantile(0.25), col.quantile(0.75)
iqr = q3 - q1
print('Q1 (25%):', round(q1, 2))
print('Q3 (75%):', round(q3, 2))
print('IQR     :', round(iqr, 2), '-> spread of the middle 50% (robust to outliers)')
```

```python
tip=df['tip']
print('Range :', round(tip.max() - tip.min(), 2))
print('Var   :', round(tip.var(), 2))
print('Std   :', round(tip.std(), 2))
q1_tip, q3_tip = tip.quantile(0.25), tip.quantile(0.75)
iqr_tip = q3_tip - q1_tip
print('Q1 (25%):', round(q1_tip, 2))
print('Q3 (75%):', round(q3_tip, 2))
print('IQR     :', round(iqr_tip, 2))
```

```python
col = df['total_bill']
five = col.describe()[['min', '25%', '50%', '75%', 'max']]
print(five)
print('\nSkewness:', round(col.skew(), 3))
```
