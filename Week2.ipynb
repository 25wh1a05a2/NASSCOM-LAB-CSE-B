# 1.1 Generate a list of 100 simulated sensor readings.
#    Each reading should be a tuple: (timestamp_str, device_id, device_type, location, value).
#    - timestamp_str: Format as 'YYYY-MM-DD HH:MM:SS'
#    - device_id: A unique ID for each reading (e.g., 'DEV001', 'DEV002', etc., up to 100 unique IDs).
#    - device_type: Randomly chosen from DEVICE_TYPES.
#    - location: Randomly chosen from LOCATIONS.
#    - value: Random float within READING_RANGES for its device_type (round to 2 decimal places).
# Store these in a list called 'raw_sensor_data'.


```python
import datetime
import os
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
DEVICE_TYPES = [
    "Temperature_Sensor",
    "Humidity_Sensor",
    "Pressure_Sensor",
    "Light_Sensor",
]
LOCATIONS = ["Lab1", "Lab2", "OfficeA", "OfficeB", "Warehouse"]
READING_RANGES = {
    "Temperature_Sensor": (18.0, 30.0),  # Celsius
    "Humidity_Sensor": (30.0, 80.0),  # Percentage
    "Pressure_Sensor": (980.0, 1030.0),  # hPa
    "Light_Sensor": (100, 1000),  # Lux
}

# --- Task 1: Data Generation and Basic Structures ---
print("\n--- Task 1: Data Generation and Basic Structures ---")

# Initialize the list to store raw data
raw_sensor_data = []

# Base time start point for generating sequential timestamps
start_time = datetime.datetime.now()

for i in range(1, 101):
    # 1. Generate unique sequential device ID (DEV001 to DEV100)
    device_id = f"DEV{i:03d}"

    # 2. Randomly select device type and location
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    # 3. Generate sequential timestamp (incrementing by 5 minutes per reading)
    current_time = start_time + datetime.timedelta(minutes=5 * i)
    timestamp_str = current_time.strftime("%Y-%m-%d %H:%M:%S")

    # 4. Generate random float value within the device's specific range
    min_val, max_val = READING_RANGES[device_type]
    value = round(random.uniform(min_val, max_val), 2)

    # 5. Append the constructed tuple to the list
    raw_sensor_data.append(
        (timestamp_str, device_id, device_type, location, value)
    )

# Print the first 5 records to verify the output structure
print("\nFirst 5 generated sensor readings:")
for reading in raw_sensor_data[:5]:
    print(reading)

# Print total length to verify 100 entries
print(f"\nTotal records generated: {len(raw_sensor_data)}")
```

#     a tuple: (min_value, max_value) observed for that device type across all readings.
#     Initialize with sensible large/small values.

```python
import datetime
import os
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")

# --- Global Data for the Exercise (Simulated Data) ---
DEVICE_TYPES = [
    "Temperature_Sensor",
    "Humidity_Sensor",
    "Pressure_Sensor",
    "Light_Sensor",
]
LOCATIONS = ["Lab1", "Lab2", "OfficeA", "OfficeB", "Warehouse"]
READING_RANGES = {
    "Temperature_Sensor": (18.0, 30.0),  # Celsius
    "Humidity_Sensor": (30.0, 80.0),  # Percentage
    "Pressure_Sensor": (980.0, 1030.0),  # hPa
    "Light_Sensor": (100, 1000),  # Lux
}


print("\n--- Running Task 1.4: Calculating Device Stats ---")

# Step 1: Initialize the dictionary with sensible extreme values
# (Infinity handles any real numerical range initialization beautifully)
device_stats = {
    device_type: (float("inf"), float("-inf")) for device_type in DEVICE_TYPES
}

# Step 2: Loop through raw data to find observed absolute min and max
for _, _, device_type, _, value in raw_sensor_data:
    current_min, current_max = device_stats[device_type]

    # Update min if current value is smaller
    if value < current_min:
        current_min = value

    # Update max if current value is larger
    if value > current_max:
        current_max = value

    # Save the updated metrics back as a tuple
    device_stats[device_type] = (current_min, current_max)

# Print the final dictionary values to verify results
for device, stats in device_stats.items():
    print(f"{device}: Min Observed = {stats[0]}, Max Observed = {stats[1]}")
```

# 2.2 Write a function `calculate_average_temperature(temp_data_list)`
#     that takes the 'temperature_readings' list (list of dictionaries).
#     It should calculate and return the average temperature. Return 0.0 if the list is empty.

```python
def calculate_average_temperature(temp_data_list):
    """
    Calculates the average temperature from a list of temperature dictionaries.

    Args:
        temp_data_list (list): A list of dictionaries, where each dictionary
                               is expected to have a 'temperature_celsius' key.

    Returns:
        float: The average temperature, or 0.0 if the list is empty.
    """
    if not temp_data_list:
        return 0.0

    total_temperature = sum(item['temperature_celsius'] for item in temp_data_list)
    return total_temperature / len(temp_data_list)

print("\n--- Running Task 2.2: Calculating Average Temperature ---")

average_temp = calculate_average_temperature(temperature_data)
print(f"Average temperature from all records: {average_temp:.2f}°C")

# Test with an empty list
empty_list_avg_temp = calculate_average_temperature([])
print(f"Average temperature from an empty list: {empty_list_avg_temp:.2f}°C")
```

# 2.3 Write a function `generate_report_line(reading_dict)`
#     that takes a single temperature reading dictionary.
#     It should return a formatted string like:
#     "[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C"

```python
def generate_report_line(reading_dict):
    """
    Generates a formatted report string for a single temperature reading dictionary.

    Args:
        reading_dict (dict): A dictionary with keys 'timestamp', 'device_id',
                             'location', and 'temperature_celsius'.

    Returns:
        str: A formatted string like "[{timestamp}] Device {device_id} at {location}: {temperature_celsius}°C".
    """
    timestamp = reading_dict['timestamp']
    device_id = reading_dict['device_id']
    location = reading_dict['location']
    temperature = reading_dict['temperature_celsius']
    return f"[{timestamp}] Device {device_id} at {location}: {temperature:.2f}°C"


print("\n--- Running Task 2.3: Generating Report Lines ---")

# Demonstrate the function with the first temperature reading
if temperature_data:
    sample_reading = temperature_data[0]
    report_line = generate_report_line(sample_reading)
    print(f"Sample Report Line: {report_line}")
else:
    print("No temperature data available to generate a sample report line.")
```

```python
# 3.2 Calculate the duration between the first and last timestamp in 'raw_sensor_data'.
#     Convert string timestamps to datetime objects first.
#     Print the duration in days, hours, and minutes.

# Your code for Task 3.2 here:
import datetime

if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    # Convert string to datetime object
    # The format string must exactly match the timestamp string
    date_format = '%Y-%m-%d %H:%M:%S'
    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()
    hours = int(seconds // 3600) % 24 # Calculate remaining hours after full days
    minutes = int((seconds % 3600) // 60) # Calculate remaining minutes after full hours

    print(f"\nDuration of data collection: {days} days, {hours} hours, {minutes} minutes.")
else:
    print("\nNot enough data to calculate duration.")
```
